# TeaBioSens exploratory analysis

This notebook keeps the 597 rating rows separate from the 30 independent recipes and inspects the finite 32-point experimental design. It does not fit models or split observations by row.

```python
from pathlib import Path
import sys
import pandas as pd
import plotly.express as px
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from tea_ai.design_space import RecipeDesignSpace, enumerate_valid_design_points
ratings = pd.read_csv(ROOT / 'data/processed/sensory_observations.csv')
blends = pd.read_csv(ROOT / 'data/processed/blend_master.csv')
len(ratings), blends['sample_code'].nunique(), len(enumerate_valid_design_points())
```

## Design-space coverage

```python
design = RecipeDesignSpace(blends)
design.describe()
```

```python
design_table = blends[['sample_code', 'green_g', 'white_g', 'oolong_g', 'black_g', 'overall_score_mean', 'overall_score_std', 'n_ratings']]
px.scatter(design_table, x='green_g', y='overall_score_mean', color='black_g', size='n_ratings', hover_data=['sample_code', 'white_g', 'oolong_g', 'overall_score_std'], title='Measured Overall score by Green / Black mass')
```
